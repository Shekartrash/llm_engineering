```python
import pprint
import boto3

# Initialize Pretty Printer
pp = pprint.PrettyPrinter(indent=2)

# Initialize Bedrock Client
bedrock = boto3.client(
    service_name='bedrock',
    region_name='eu-central-1')  # Adjust if needed

def list_foundation_models():
    models = bedrock.list_foundation_models()
    
    for model in models["modelSummaries"]:
        pp.pprint(model)  # Pretty print model details
        print("-" * 30)  # Print separator

# Call the function
list_foundation_models()
```

```python
def get_foundation_model(modelIdentifier):
    model = bedrock.get_foundation_model(modelIdentifier=modelIdentifier)
    pp.pprint(model)
```
